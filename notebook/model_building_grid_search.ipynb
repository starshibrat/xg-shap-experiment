```python
import os
import sys
module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)
import pandas as pd
from utils.cleaning import *
from tree.custom_xgboost import CustomXGBClassifier
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from common.helper import load_data_and_scale_pos_weight
from common.score import pr_auc_func
from sklearn.metrics import make_scorer
```

```python
X_train, _, y_train, _, _ = load_data_and_scale_pos_weight(cast_string=True)
```

```python
param_grid = {
    'learning_rate': [0.05, 0.1, 0.5],
    'max_depth': [2, 3, 4, 5],
    'n_estimators': [50, 100, 200, 300],
}
```

```python
cv = StratifiedKFold(
    n_splits=5,
)
```

```python
xgb = CustomXGBClassifier()
scoring = make_scorer(
    pr_auc_func,
    response_method='predict_proba',
    greater_is_better=True
)

grid_search = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid,
    cv=cv,
    n_jobs=-1,
    verbose=2,
    scoring=scoring,
    
)

grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
grid_search.cv_results_
```

```python
results_df = pd.DataFrame(grid_search.cv_results_)
results_df = results_df.sort_values(by=["rank_test_score"])
results_df = results_df.set_index(
    results_df["params"].apply(lambda x: "_".join(str(val) for val in x.values()))
).rename_axis("kernel")
results_df[["params", "rank_test_score", "mean_test_score", "std_test_score"]]
```
